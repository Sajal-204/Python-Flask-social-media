```python
# Night two as of sep 26 / 2026
# memory recall

class Bank:
  def __init__(self,holder,balance=0):
    self.holder_name=holder
    self.balance_amount=balance

  def deposit(self,amount):
    self.balance_amount+=amount
    print(f" The new balance is {self.balance_amount}")
  def withdraw(self,amount):
    if amount>self.balance_amount:
      print("sorry!! withdraw amount is more than actual balance ")
    else:
      self.balance_amount-=amount
      print(f"withdrawl of {amount}")
  def balance(self):
    print(f" the balance is {self.balance_amount}")

my_ac=Bank("sajal",15000)
my_ac.deposit(5000)
my_ac.withdraw(10000)
my_ac.balance()
```

```python
# Night two as of sep 26 / 2026
# Inheritance of oop
class User:

  def __init__(self,username,email):
    self.username=username
    self.email=email

  def get_details(self):
    print(f" User : {self.username} | Email : {self.email}")

class AdminUser(User):
  def __init__(self, username, email,permisins_level):
    super().__init__(username,email) # it directly takes the value of main class
    self.permision=permisins_level

  def show_permissions(self):
    print(f" {self.username} has {self.permision} access. ")

user1=User("alex","alex@.com")
user1.get_details()

user2=AdminUser("sam","sam@.com","admin")
user2.get_details()
user2.show_permissions()

```
